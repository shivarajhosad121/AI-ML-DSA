```python
#Finding largest element
arr = [4, 7, 1, 9, 3]
largest = arr[0]
for i in arr:
    if i > largest:
        largest=i

print(largest)
        
```

```python
#finding smallest largest element
arr = [4, 7, 1, 9, 3]
smallest = arr[0]
for i in arr:
    if i < smallest:
        smallest=i

print(smallest)
        
```

```python
#finding second smallest
arr = [9, 9, 8, 7, 6]

smallest = arr[0]
s_smallest = None

for i in arr:
    if i < smallest:
        s_smallest = smallest
        smallest = i

    elif i > smallest and (s_smallest is None or i < s_smallest):
        s_smallest = i

print(s_smallest)
```
