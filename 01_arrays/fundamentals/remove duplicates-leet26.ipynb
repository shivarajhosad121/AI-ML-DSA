```python
#array insertion
def insert_at(arr, index, value):
    arr.append(0)  # create space at the end

    for i in range(len(arr) - 1, index, -1):
        arr[i] = arr[i - 1]

    arr[index] = value


arr = [10, 20, 30, 40, 50]

insert_at(arr, 2, 25)

print(arr)
        
```

```python
#array deletion
def delete_at(arr, index):

    for i in range(index, len(arr) - 1):
        arr[i] = arr[i + 1]

    arr.pop()

    return arr


arr = [10, 20, 30, 40, 50]

print(delete_at(arr, 2))
    
```

```python
#LeetCode problem: LeetCode 26 — Remove Duplicates from Sorted Array.
def remove_duplicates(arr):
    if len(arr) == 0:
        return 0

    i = 0

    for j in range(1, len(arr)):
        if arr[j] != arr[i]:
            i += 1
            arr[i] = arr[j]

    return i + 1


arr = [1, 1, 2, 2, 3, 3, 4]

k = remove_duplicates(arr)

print(k)
print(arr[:k])
```
