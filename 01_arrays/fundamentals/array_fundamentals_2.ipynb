```python
def count_occurrences(arr, target):
    count = 0

    for i in range(len(arr)):
        if arr[i] == target:
            count += 1

    return count


arr = [4, 7, 2, 7, 9, 7, 3]
target = 7

print(count_occurrences(arr, target))
```

```python
def find_all_indices(arr, target):
    indices = []

    for i in range(len(arr)):
        if arr[i] == target:
            indices.append(i)

    return indices


arr = [4, 7, 2, 7, 9, 7, 3]
target = 7

print(find_all_indices(arr, target))
```

```python
#reversing an array using two-pointer pattern
def reverse_array(arr):
    left = 0
    right = len(arr) - 1

    while left < right:
        arr[left], arr[right] = arr[right], arr[left]

        left += 1
        right -= 1

    return arr


arr = [4, 7, 1, 9, 3]

print(reverse_array(arr))
```

```python
def move_zeroes(arr):
    left = 0

    for i in range(len(arr)):
        if arr[i] != 0:
            arr[left], arr[i] = arr[i], arr[left]
            left += 1

    return arr


arr = [0, 1, 0, 3, 12]

print(move_zeroes(arr))
```
